## Vector Operations

```python
u = np.array([2, 4, 5, 6])
```

```python
u
```

```python
v = np.array([1, 0, 0, 2])
```

```python
v
```

```python
u + v
```

```python
u * v
```

```python
def vect_vect_mult(u,v):
    assert u.shape[0] == v.shape[0]
    n = u.shape[0]
    result = 0.0
    for i in range(n):
        result += u[i] * v[i]
    return result
```

```python
vect_vect_mult(u,v)
```

```python
u.dot(v)
```

```python
u
```

```python
U = np.array([
    [2, 4, 5, 6],
    [1, 2, 1, 2],
    [3, 1, 2, 1]
])
```

```python
U
```

```python
v = np.array([1, 0.5, 2, 1])
```

```python
v
```

```python
def mat_vec_mult(U,v):
    assert U.shape[1] == v.shape[0]
    num_rows = U.shape[0]
    result = np.zeros(num_rows)
    for i in range(num_rows):
        result[i] = vect_vect_mult(U[i],v)
    return result
```

```python
mat_vec_mult(U,v)
```

```python
U.dot(v)
```

```python
V = np.array([
    [1, 1, 2],
    [0, 0.5, 1],
    [0, 2, 1],
    [2, 1, 0]
])
```

```python
V
```

```python
def mat_mat_mult(U,V):
    assert U.shape[1] == V.shape[0]
    num_rows = U.shape[0]
    num_cols = V.shape[1]
    result = np.zeros((num_rows, num_cols))
    for j in range(num_cols):
        vj = V[:, j]
        Uvj = mat_vec_mult(U, vj)
        result[:, j] = Uvj
    return result
```

```python
mat_mat_mult(U, V)
```

```python
U.dot(V)
```
