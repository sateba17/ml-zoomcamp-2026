## Multiplication

```python
u.shape[0]
```

```python
def vect_vect_mult(u, v):
    assert u.shape[0] == v.shape[0]

    n = u.shape[0]

    result = 0.0

    for i in range(n):
        result += u[i] * v[i]

    return result
```

```python
vect_vect_mult(u,v)
```

```python
u.dot(v)
```

```python
u.dot(v)+10
```

```python
U = np.array([
    [2, 4, 5, 6],
    [1, 2, 1, 2],
    [3, 1, 2, 1]
])
```

```python
U
```

```python
U.shape
```

```python
U[0]
```

```python
def mat_vect_mult(U,v):
    assert U.shape[1] == v.shape[0]

    num_rows = U.shape[0]
    result = np.zeros(num_rows)

    for i in range(num_rows):
        result[i] = vect_vect_mult(U[i], v)
    return result
```

```python
mat_vect_mult(U,v)
```

## Matrix Matrix Multiplication

```python
V = np.array([
    [1, 1, 2],
    [0, 0.5, 1],
    [0, 2, 1],
    [2, 1, 0]
])
```

```python
def mat_mat_mult(U,V):
    assert U.shape[1] == V.shape[0]

    num_rows = U.shape[0]
    num_cols = V.shape[1]

    result = np.zeros((num_rows, num_cols))

    for i in range(num_rows):
        vi = V[:,i]
        result[:,i] = mat_vect_mult(U,vi)
    return result
```

```python
mat_mat_mult(U,V)
```

```python
U.dot(V)
```
