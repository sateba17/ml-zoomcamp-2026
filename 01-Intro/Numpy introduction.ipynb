## Randomly Generated Arrays

```python
np.random.seed(2)
np.random.rand(5, 2)
```

```python
np.random.seed(2)
np.random.randn(5,2)
```

```python
np.random.seed(2)
np.random.randint(low=0, high=100, size=(5,2))
```
